```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.columns
```

ubah label education menjadi teks

```python
labels = {1: 'Undergraduate', 2: 'Graduate', 3: 'Professional'}
df['education_group'] = df['education'].map(labels)
df[['education', 'education_group']].head(10)
```

Segmentasi Umur dan Pendapatan

```python
labels_age = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels_age)
df[['age', 'age_group']].head(10)
```

```python
labels_income = ['Low Income', 'Medium Income', 'High Income']
df['income_group'] = pd.qcut(df['income'], q=3, labels = labels_income)
df[['income', 'income_group']].head(10)
```

Status KPR memisahkan yang punya dan tidak punta

```python
df['has_mortgage'] = df['mortgage'].apply(lambda x: 'Has Mortgage' if x > 0 else 'No Mortgage')
df[['mortgage', 'has_mortgage']].head(10)
```
